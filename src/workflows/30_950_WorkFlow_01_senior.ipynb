```r
# Escrito por alumnos de  Universidad Austral  Rosario

Corregir_MICE <- function(pcampo, pmeses) {

  meth <- rep("", ncol(dataset))
  names(meth) <- colnames(dataset)
  meth[names(meth) == pcampo] <- "sample"

  # llamada a mice  !
  imputacion <- mice(dataset,
    method = meth,
    maxit = 5,
    m = 1,
    seed = 7)

  tbl <- mice::complete(dataset)

  dataset[, paste0(pcampo) := ifelse(foto_mes %in% pmeses, tbl[, get(pcampo)], get(pcampo))]

}
```

```r
Corregir_interpolar <- function(pcampo, pmeses) {

  tbl <- dataset[, list(
    "v1" = shift(get(pcampo), 1, type = "lag"),
    "v2" = shift(get(pcampo), 1, type = "lead")
  ),
  by = eval(envg$PARAM$dataset_metadata$entity_id)
  ]

  tbl[, paste0(envg$PARAM$dataset_metadata$entity_id) := NULL]
  tbl[, promedio := rowMeans(tbl, na.rm = TRUE)]

  dataset[
    ,
    paste0(pcampo) := ifelse(!(foto_mes %in% pmeses),
      get(pcampo),
      tbl$promedio
    )
  ]
}
```

```r
drift_rank_simple <- function(campos_drift) {

  cat( "inicio drift_rank_simple()\n")
  for (campo in campos_drift)
  {
    cat(campo, " ")
    dataset[, paste0(campo, "_rank") :=
      (frank(get(campo), ties.method = "random") - 1) / (.N - 1), by = list(foto_mes)]
    dataset[, (campo) := NULL]
  }
  cat( "fin drift_rank_simple()\n")
}
```

```r
# El cero se transforma en cero
# los positivos se rankean por su lado
# los negativos se rankean por su lado

drift_rank_cero_fijo <- function(campos_drift) {

  cat( "inicio drift_rank_cero_fijo()\n")
  for (campo in campos_drift)
  {
    cat(campo, " ")
    dataset[get(campo) == 0, paste0(campo, "_rank") := 0]
    dataset[get(campo) > 0, paste0(campo, "_rank") :=
      frank(get(campo), ties.method = "random") / .N, by = list(foto_mes)]

    dataset[get(campo) < 0, paste0(campo, "_rank") :=
      -frank(-get(campo), ties.method = "random") / .N, by = list(foto_mes)]
    dataset[, (campo) := NULL]
  }
  cat("\n")
  cat( "fin drift_rank_cero_fijo()\n")
}
```

```r
drift_estandarizar <- function(campos_drift) {

  cat( "inicio drift_estandarizar()\n")
  for (campo in campos_drift)
  {
    cat(campo, " ")
    dataset[, paste0(campo, "_normal") :=
      (get(campo) -mean(campo, na.rm=TRUE)) / sd(get(campo), na.rm=TRUE),
      by = list(foto_mes)]

    dataset[, (campo) := NULL]
  }
  cat( "fin drift_estandarizar()\n")
}
```

#### 9.7.1.4  FEhist Feature Engineering historico

El Fature Engineering Histórico es la etapa que más aporta a la ganancia final, ya que enriquece cada registro del dataset con su historia.

Para cada campo del dataset original (*)
se crean lo siguientes campos de a partir de la historia
* lag1  lags de orden 1
* delta1  =  valor actual - lag1
* lag2  lags de orden 2
* delta2  = valor actual - lag2


(*) Excepto para los campos  <numero_de_cliente,  foto_mes,  clase_ternaria>

```r
if( !require("Rcpp")) install.packages("Rcpp", repos = "http://cran.us.r-project.org")
require("Rcpp")
```

```r
# se calculan para los 6 meses previos el minimo, maximo y
#  tendencia calculada con cuadrados minimos
# la formula de calculo de la tendencia puede verse en
#  https://stats.libretexts.org/Bookshelves/Introductory_Statistics/Book%3A_Introductory_Statistics_(Shafer_and_Zhang)/10%3A_Correlation_and_Regression/10.04%3A_The_Least_Squares_Regression_Line
# para la maxíma velocidad esta funcion esta escrita en lenguaje C,
# y no en la porqueria de R o Python

cppFunction("NumericVector fhistC(NumericVector pcolumna, IntegerVector pdesde )
{
  /* Aqui se cargan los valores para la regresion */
  double  x[100] ;
  double  y[100] ;

  int n = pcolumna.size();
  NumericVector out( 5*n );

  for(int i = 0; i < n; i++)
  {
    //lag
    if( pdesde[i]-1 < i )  out[ i + 4*n ]  =  pcolumna[i-1] ;
    else                   out[ i + 4*n ]  =  NA_REAL ;


    int  libre    = 0 ;
    int  xvalor   = 1 ;

    for( int j= pdesde[i]-1;  j<=i; j++ )
    {
       double a = pcolumna[j] ;

       if( !R_IsNA( a ) )
       {
          y[ libre ]= a ;
          x[ libre ]= xvalor ;
          libre++ ;
       }

       xvalor++ ;
    }

    /* Si hay al menos dos valores */
    if( libre > 1 )
    {
      double  xsum  = x[0] ;
      double  ysum  = y[0] ;
      double  xysum = xsum * ysum ;
      double  xxsum = xsum * xsum ;
      double  vmin  = y[0] ;
      double  vmax  = y[0] ;

      for( int h=1; h<libre; h++)
      {
        xsum  += x[h] ;
        ysum  += y[h] ;
        xysum += x[h]*y[h] ;
        xxsum += x[h]*x[h] ;

        if( y[h] < vmin )  vmin = y[h] ;
        if( y[h] > vmax )  vmax = y[h] ;
      }

      out[ i ]  =  (libre*xysum - xsum*ysum)/(libre*xxsum -xsum*xsum) ;
      out[ i + n ]    =  vmin ;
      out[ i + 2*n ]  =  vmax ;
      out[ i + 3*n ]  =  ysum / libre ;
    }
    else
    {
      out[ i       ]  =  NA_REAL ;
      out[ i + n   ]  =  NA_REAL ;
      out[ i + 2*n ]  =  NA_REAL ;
      out[ i + 3*n ]  =  NA_REAL ;
    }
  }

  return  out;
}")
```

```r
# calcula la tendencia de las variables cols de los ultimos 6 meses
# la tendencia es la pendiente de la recta que ajusta por cuadrados minimos
# La funcionalidad de ratioavg es autoria de  Daiana Sparta,  UAustral  2021

TendenciaYmuchomas <- function(
    dataset, cols, ventana = 6, tendencia = TRUE,
    minimo = TRUE, maximo = TRUE, promedio = TRUE,
    ratioavg = FALSE, ratiomax = FALSE) {
  gc(verbose= FALSE)
  # Esta es la cantidad de meses que utilizo para la historia
  ventana_regresion <- ventana

  last <- nrow(dataset)

  # creo el vector_desde que indica cada ventana
  # de esta forma se acelera el procesamiento ya que lo hago una sola vez
  vector_ids <- dataset[ , numero_de_cliente ]

  vector_desde <- seq(
    -ventana_regresion + 2,
    nrow(dataset) - ventana_regresion + 1
  )

  vector_desde[1:ventana_regresion] <- 1

  for (i in 2:last) {
    if (vector_ids[i - 1] != vector_ids[i]) {
      vector_desde[i] <- i
    }
  }
  for (i in 2:last) {
    if (vector_desde[i] < vector_desde[i - 1]) {
      vector_desde[i] <- vector_desde[i - 1]
    }
  }

  for (campo in cols) {
    nueva_col <- fhistC(dataset[, get(campo)], vector_desde)

    if (tendencia) {
      dataset[, paste0(campo, "_tend", ventana) :=
        nueva_col[(0 * last + 1):(1 * last)]]
    }

    if (minimo) {
      dataset[, paste0(campo, "_min", ventana) :=
        nueva_col[(1 * last + 1):(2 * last)]]
    }

    if (maximo) {
      dataset[, paste0(campo, "_max", ventana) :=
        nueva_col[(2 * last + 1):(3 * last)]]
    }

    if (promedio) {
      dataset[, paste0(campo, "_avg", ventana) :=
        nueva_col[(3 * last + 1):(4 * last)]]
    }

    if (ratioavg) {
      dataset[, paste0(campo, "_ratioavg", ventana) :=
        get(campo) / nueva_col[(3 * last + 1):(4 * last)]]
    }

    if (ratiomax) {
      dataset[, paste0(campo, "_ratiomax", ventana) :=
        get(campo) / nueva_col[(2 * last + 1):(3 * last)]]
    }
  }
}
```

```r
AgregaVarRandomForest <- function() {

  cat( "inicio AgregaVarRandomForest()\n")
  gc(verbose= FALSE)
  dataset[, clase01 := 0L ]
  dataset[ clase_ternaria %in% PARAM$FE_rf$train$clase01_valor1,
      clase01 := 1L ]

  campos_buenos <- setdiff(
    colnames(dataset),
    c( "clase_ternaria", "clase01")
  )

  dataset[, entrenamiento :=
    as.integer( foto_mes %in% PARAM$FE_rf$train$training )]

  dtrain <- lgb.Dataset(
    data = data.matrix(dataset[entrenamiento == TRUE, campos_buenos, with = FALSE]),
    label = dataset[entrenamiento == TRUE, clase01],
    free_raw_data = FALSE
  )

  modelo <- lgb.train(
     data = dtrain,
     param = PARAM$FE_rf$lgb_param,
     verbose = -100
  )

  cat( "Fin construccion RandomForest\n" )
  # grabo el modelo, achivo .model
  lgb.save(modelo, file="modelo.model" )

  qarbolitos <- copy(PARAM$FE_rf$lgb_param$num_iterations)

  periodos <- dataset[ , unique( foto_mes ) ]

  for( periodo in  periodos )
  {
    cat( "periodo = ", periodo, "\n" )
    datamatrix <- data.matrix(dataset[ foto_mes== periodo, campos_buenos, with = FALSE])

    cat( "Inicio prediccion\n" )
    prediccion <- predict(
        modelo,
        datamatrix,
        type = "leaf"
    )
    cat( "Fin prediccion\n" )

    for( arbolito in 1:qarbolitos )
    {
       cat( arbolito, " " )
       hojas_arbol <- unique(prediccion[ , arbolito])

       for (pos in 1:length(hojas_arbol)) {
         # el numero de nodo de la hoja, estan salteados
         nodo_id <- hojas_arbol[pos]
         dataset[ foto_mes== periodo, paste0(
            "rf_", sprintf("%03d", arbolito),
             "_", sprintf("%03d", nodo_id)
          ) :=  as.integer( nodo_id == prediccion[ , arbolito]) ]

       }

       rm( hojas_arbol )
    }
    cat( "\n" )

    rm( prediccion )
    rm( datamatrix )
    gc(verbose= FALSE)
  }

  gc(verbose= FALSE)

  # borro clase01 , no debe ensuciar el dataset
  dataset[ , clase01 := NULL ]

}
```

```r
###
# Funciones auxiliares para separar grupos de variables
is_bool_column <- function(columname) {
  tryCatch({
    x <- dataset[[columname]]
    if (!(is.numeric(x) || is.integer(x))) return(FALSE)
    maxx <- suppressWarnings(max(x, na.rm = TRUE))
    minn <- suppressWarnings(min(x, na.rm = TRUE))
    isTRUE((maxx - minn) == 1)
  }, error = function(e) FALSE)
}

get_non_bool_cols <- function(dataset_in) {
  all_cols <- names(dataset_in)
  bool_cols <- sapply(all_cols, is_bool_column)
  all_cols[!bool_cols]
}

campos_a_omitir = c('numero_de_cliente','foto_mes','azar','clase_ternaria')
campos_monetarios <- colnames(dataset)
campos_rf <- colnames(dataset)
campos_lag <- colnames(dataset)
campos_int <- colnames(dataset)

actualizar_campos = function(){
  cm <- ci <- cl <- cr <- colnames(dataset)

  # monetarios
  cm  <- cm[cm %like% "^(m|Visa_m|Master_m|vm_m)"]

  # rf
  cr  <- cr[cr %like% "^rf"]

  # lags / deltas (regex más amplia y sin el '.$' que te recortaba)
  cl  <- cl[cl %like% "(lag|delta)"]
  cl  <- setdiff(cl, cm)

  # intermedias: sin omitidos, sin monetarios/lag/rf y sin booleanas
  ci  <- setdiff(ci, c(campos_a_omitir, cm, cl, cr))
  ci  <- ci[!sapply(ci, is_bool_column)]
  ci  <- na.omit(ci)

  # asignación global coherente
  campos_monetarios <<- cm
  campos_rf        <<- cr
  campos_lag       <<- cl
  campos_int       <<- ci
}
actualizar_campos()


###
```

```r
###
# Funciones auxiliares para la creacion de 'columnas geneticas'
# (mantiene espíritu: eval por string sobre dataset; sólo robustez extra)

.to_num <- function(x){
  if (is.data.frame(x)) x <- x[[1]]
  if (is.matrix(x))     x <- x[,1]
  if (is.list(x))       x <- unlist(x, recursive = TRUE, use.names = FALSE)
  x <- suppressWarnings(as.numeric(x))
  x[!is.finite(x)] <- NA_real_
  x
}

safeEval <- function(expr) {
  val <- try(eval(parse(text = expr), envir = dataset), silent = TRUE)
  if (inherits(val, "try-error")) return(rep(NA_real_, nrow(dataset)))
  val <- .to_num(val)
  if (length(val) != nrow(dataset)) return(rep(NA_real_, nrow(dataset)))
  val
}

count_operators <- function(expr) {
  text <- paste(deparse(expr), collapse = " ")
  stringr::str_count(text, "\\+|-|/|\\*")
}

clean_column_name_base <- function(expr_str) {
  txt <- as.character(parse(text = expr_str))
  txt <- gsub("dataset|\\[|\\]|\"", "", txt)
  txt <- gsub(",\\s*", "", txt)
  txt <- gsub("[^A-Za-z0-9_]+", "_", txt)
  trimws(substr(txt, 1, 80))
}

# ==== Unarios "baratos" y estables (NUEVOS) ====
abs_       <- function(x) { x <- .to_num(x); abs(x) }
square     <- function(x) { x <- .to_num(x); x * x }
log1p_pos  <- function(x) { x <- .to_num(x); log1p(pmax(x, 0)) }
sqrt_pos   <- function(x) { x <- .to_num(x); sqrt(pmax(x, 0)) }
relu_      <- function(x) { x <- .to_num(x); pmax(x, 0) }
neg_relu_  <- function(x) { x <- .to_num(x); pmax(-x, 0) }
sign_      <- function(x) { x <- .to_num(x); sign(x) }

clip_p01   <- function(x) {
  x <- .to_num(x)
  q <- suppressWarnings(quantile(x, c(0.01, 0.99), na.rm = TRUE))
  pmin(pmax(x, q[1]), q[2])
}

# ==== Binarios simples (NUEVOS) ====
pmin_ <- function(a, b) { a <- .to_num(a); b <- .to_num(b); pmin(a, b) }
pmax_ <- function(a, b) { a <- .to_num(a); b <- .to_num(b); pmax(a, b) }

###
```

```r
###
# funcion que se busca optimizar con las variables. El algoritmo minimiza, por lo que el return es 1 - AUC.
fitness_function <- function(expr) {
  feat <- safeEval(expr)

  # guardas rápidas
  if (!is.numeric(feat) || sum(is.finite(feat)) < 100) return(3)
  # si es constante, no aporta
  rng <- suppressWarnings(range(feat, na.rm = TRUE))
  if (is.finite(rng[1]) && is.finite(rng[2]) && (rng[2] - rng[1] == 0)) return(3)

  d <- data.frame(y = dataset$clase01, x = feat)
  colnames(d) <- c('y','x')
  if (length(unique(na.omit(d$y))) < 2) return(3)

  dtrain <- lgb.Dataset(data.matrix(d["x"]), label = d$y)

  idx <- sample(seq_len(nrow(d)), size = max(1L, floor(0.2 * nrow(d))))
  dvalid <- lgb.Dataset(data.matrix(d[idx, "x", drop=FALSE]), label = d$y[idx])
  valids <- list(valid = dvalid)

  res <- tryCatch({
    model <- lgb.train(
      params = list(
        objective = "binary",
        metric = "auc",
        is_unbalance = TRUE,
        verbosity = -1,
        min_data_in_leaf = 5,
        num_leaves = 15
      ),
      data = dtrain,
      nrounds = 20,
      valids = valids,
      record = TRUE
    )
    1 - tail(unlist(model$record_evals$valid$auc$eval), 1)
  }, error = function(e) 3)
  res
}

###
```

```r
# explicado en el video

grammarDef <- CreateGrammar(list(
  expr = gsrule(
      # === TUS REGLAS ORIGINALES (NO TOCO) ===
      "dataset[,<m_var>]/dataset[,<m_var>]",
      "dataset[,<m_var>]/dataset[,<i_var>]",
      "dataset[,<m_var>]/dataset[,<l_var>]",
      "dataset[,<m_var>]+dataset[,<l_var>]",
      "dataset[,<m_var>]-dataset[,<l_var>]",
      "dataset[,<l_var>]/dataset[,<m_var>]",
      "dataset[,<l_var>]/dataset[,<i_var>]",
      "dataset[,<l_var>]/dataset[,<l_var>]",
      "dataset[,<l_var>]+dataset[,<l_var>]",
      "dataset[,<l_var>]-dataset[,<l_var>]",
      "dataset[,<i_var>]/dataset[,<i_var>]",
      "dataset[,<i_var>]*dataset[,<i_var>]",
      "dataset[,<rf_var>]+dataset[,<rf_var>]",

      # === ADICIONALES LIVIANOS (DIVERSIDAD) ===

      # Unarios sobre monetarios / internos / lags / rf
      "abs_(dataset[,<m_var>])",
      "abs_(dataset[,<l_var>])",
      "log1p_pos(dataset[,<m_var>])",
      "log1p_pos(dataset[,<i_var>])",
      "sqrt_pos(dataset[,<m_var>])",
      "sqrt_pos(dataset[,<i_var>])",
      "square(dataset[,<i_var>])",
      "clip_p01(dataset[,<m_var>])",
      "relu_(dataset[,<l_var>])",
      "neg_relu_(dataset[,<l_var>])",
      "sign_(dataset[,<l_var>])",

      # Combinadores simples con lags
      "pmin_(dataset[,<m_var>], dataset[,<l_var>])",
      "pmax_(dataset[,<m_var>], dataset[,<l_var>])",
      "pmin_(dataset[,<i_var>], dataset[,<l_var>])",
      "pmax_(dataset[,<i_var>], dataset[,<l_var>])",

      # Interacciones con RF (muy básicas)
      "dataset[,<rf_var>]*dataset[,<l_var>]",
      "dataset[,<rf_var>]/dataset[,<m_var>]",
      "dataset[,<rf_var>]-dataset[,<i_var>]",

      # Lags “vs” monetarios: diferencias/ratios estabilizados
      "log1p_pos(dataset[,<l_var>]) - log1p_pos(dataset[,<m_var>])",
      "sqrt_pos(dataset[,<l_var>]) + sqrt_pos(dataset[,<i_var>])",
      "abs_(dataset[,<l_var>]) / (1 + abs_(dataset[,<m_var>]))",

      # Entre lags e internos (ligeramente no lineal)
      "dataset[,<l_var>]*dataset[,<l_var>]",
      "dataset[,<l_var>]*dataset[,<i_var>]"
      ),
  m_var  = gvrule(setdiff(campos_monetarios,c("clase_ternaria","clase01","foto_mes",'azar'))),
  i_var  = gvrule(setdiff(campos_int,c("clase_ternaria","clase01","foto_mes",'azar'))),
  l_var  = gvrule(setdiff(campos_lag,c("clase_ternaria","clase01","foto_mes",'azar'))),
  rf_var = gvrule(setdiff(campos_rf,c("clase_ternaria","clase01","foto_mes",'azar')))
  )
)


###
```

```r
###
# Funcion principal
agregarVariablesGeneticas <- function(){
  dataset[, clase01 := ifelse( clase_ternaria %in% c("BAJA+1","BAJA+2"), 1, 0 )]

  for (var in 1:PARAM$FE_GENALG$new_variables){

    actualizar_campos()
    ge_result <- GrammaticalEvolution(
      grammarDef,
      fitness_function,
      terminationCost = 0.7,
      max.depth = PARAM$FE_GENALG$max_depth,
      iterations = PARAM$FE_GENALG$generations,
      popSize = PARAM$FE_GENALG$population
    )

    best_expr <- ge_result$best$expression

    # generamos la columna nueva (vector numérico, mismo largo que dataset)
    new_feature <- safeEval(best_expr)

    # nombre limpio (y único si hiciera falta)
    coln <- clean_column_name_base(best_expr)
    if (coln %in% names(dataset)) {
      suf <- 1L
      while (paste0(coln, "_", suf) %in% names(dataset)) suf <- suf + 1L
      coln <- paste0(coln, "_", suf)
    }

    # asignación directa a data.table
    dataset[, (coln) := new_feature]
  }
  dataset[, clase01 := NULL]
  gc(verbose=FALSE)
}

###
```

```r
VPOS_CORTE <- c()

fganancia_lgbm_meseta <- function(probs, datos) {
  vlabels <- get_field(datos, "label")
  vpesos <- get_field(datos, "weight")

  tbl <- as.data.table(list(
    "prob" = probs,
    "gan" = ifelse(vlabels == 1 & vpesos > 1, PARAM$CN$train$gan1, PARAM$CN$train$gan0)
  ))

  setorder(tbl, -prob)
  tbl[, posicion := .I]
  tbl[, gan_acum := cumsum(gan)]
  setorder(tbl, -gan_acum) # voy por la meseta

  gan <- mean(tbl[1:500, gan_acum]) # meseta de tamaño 500

  pos_meseta <- tbl[1:500, median(posicion)]
  VPOS_CORTE <<- c(VPOS_CORTE, pos_meseta)

  return(list(
    "name" = "ganancia",
    "value" = gan,
    "higher_better" = TRUE
  ))
}
```

```r
# Elimina del dataset las variables que estan por debajo
#  de la capa geologica de canaritos
# se llama varias veces, luego de agregar muchas variables nuevas,
#  para ir reduciendo la cantidad de variables
# y así hacer lugar a nuevas variables importantes

GVEZ <- 1

campitos <- c( "numero_de_cliente", "foto_mes", "clase_ternaria" )

CanaritosAsesinos <- function(
  canaritos_ratio,
  canaritos_desvios,
  canaritos_semilla) {

  cat( "inicio CanaritosAsesinos()\n")
  gc(verbose= FALSE)
  dataset[, clase01 := 0L ]
  dataset[ clase_ternaria %in% PARAM$CN$train$clase01_valor1,
      clase01 := 1L ]

  set.seed(canaritos_semilla, kind = "L'Ecuyer-CMRG")
  for (i in 1:(ncol(dataset) * canaritos_ratio)) {
    dataset[, paste0("canarito", i) := runif(nrow(dataset))]
  }

  campos_buenos <- setdiff(
    colnames(dataset),
    c( campitos, "clase01")
  )

  azar <- runif(nrow(dataset))

  dataset[, entrenamiento :=
    as.integer( foto_mes %in% PARAM$CN$train$training &
      (clase01 == 1 | azar < PARAM$CN$train$undersampling))]

  dtrain <- lgb.Dataset(
    data = data.matrix(dataset[entrenamiento == TRUE, campos_buenos, with = FALSE]),
    label = dataset[entrenamiento == TRUE, clase01],
    weight = dataset[
      entrenamiento == TRUE,
      ifelse(clase_ternaria %in% PARAM$CN$train$positivos, 1.0000001, 1.0)
    ],
    free_raw_data = FALSE
  )

  dvalid <- lgb.Dataset(
    data = data.matrix(dataset[foto_mes %in% PARAM$CN$train$validation, campos_buenos, with = FALSE]),
    label = dataset[foto_mes %in% PARAM$CN$train$validation, clase01],
    weight = dataset[
      foto_mes %in% PARAM$CN$train$validation,
      ifelse( clase_ternaria %in% PARAM$CN$train$positivos, 1.0000001, 1.0)
    ],
    free_raw_data = FALSE
  )


  param <- list(
    objective = "binary",
    metric = "custom",
    first_metric_only = TRUE,
    boost_from_average = TRUE,
    feature_pre_filter = FALSE,
    verbosity = -100,
    seed = canaritos_semilla,
    max_depth = -1, # -1 significa no limitar,  por ahora lo dejo fijo
    min_gain_to_split = 0.0, # por ahora, lo dejo fijo
    lambda_l1 = 0.0, # por ahora, lo dejo fijo
    lambda_l2 = 0.0, # por ahora, lo dejo fijo
    max_bin = 31, # por ahora, lo dejo fijo
    num_iterations = 9999, # un numero grande, lo limita early_stopping_rounds
    force_row_wise = TRUE, # para que los alumnos no se atemoricen con  warning
    learning_rate = 0.065,
    feature_fraction = 1.0, # lo seteo en 1
    min_data_in_leaf = 260,
    num_leaves = 60,
    early_stopping_rounds = 200,
    num_threads = 1
  )

  set.seed(canaritos_semilla, kind = "L'Ecuyer-CMRG")
  modelo <- lgb.train(
    data = dtrain,
    valids = list(valid = dvalid),
    eval = fganancia_lgbm_meseta,
    param = param,
    verbose = -100
  )

  tb_importancia <- lgb.importance(model = modelo)
  tb_importancia[, pos := .I]

  fwrite(tb_importancia,
    file = paste0("impo_", GVEZ, ".txt"),
    sep = "\t"
  )

  GVEZ <<- GVEZ + 1

  umbral <- tb_importancia[
    Feature %like% "canarito",
    median(pos) + canaritos_desvios * sd(pos)
  ] # Atencion corto en la mediana mas desvios!!

  col_utiles <- tb_importancia[
    pos < umbral & !(Feature %like% "canarito"),
    Feature
  ]

  col_utiles <- unique(c(
    col_utiles,
    c(campitos, "mes")
  ))

  col_inutiles <- setdiff(colnames(dataset), col_utiles)

  dataset[, (col_inutiles) := NULL]

  cat( "fin CanaritosAsesinos()\n")

  return( tb_importancia )
}
```

Nuevos  pseudo hiperparámetros

*   leaf_size (0.0,  0.5]  es el   min_data_in_leaf / nrow(dataset)

```r
# === CHUNK 1: ParamSets para OB y Semillero (invariantes al tamaño) ===
# Pseudo-HP: leaf_size (ratio) y coverage → se traducen a min_data_in_leaf y num_leaves

if (!exists("PARAM")) PARAM <- list()
if (is.null(PARAM$hiperparametertuning)) PARAM$hiperparametertuning <- list()

# --- ParamSet AMPLIO para la OB (exploración real) ---
PARAM$hiperparametertuning$hs_full <- makeParamSet(
  makeIntegerParam("num_iterations",    lower = 50L,  upper = 800L),
  makeNumericParam("learning_rate",     lower = 0.02, upper = 0.20),
  makeNumericParam("feature_fraction",  lower = 0.40, upper = 0.95),
  makeNumericParam("coverage",          lower = 0.02, upper = 0.50),
  makeNumericParam("leaf_size",         lower = 0.0005, upper = 0.020),
  makeNumericParam("bagging_fraction",  lower = 0.50, upper = 0.90),
  makeIntegerParam("bagging_freq",      lower = 1L, upper = 1L)
)

# --- ParamSet ESPECÍFICO (zona "débil") para el semillero ---
nrows_for_ratio <- if (exists("dtrain")) nrow(dtrain) else nrow(dataset)
leaf_lo_ratio <- max(300,  ceiling(0.001 * nrows_for_ratio)) / nrows_for_ratio
leaf_hi_ratio <- min(2000, ceiling(0.200 * nrows_for_ratio)) / nrows_for_ratio
leaf_lo_ratio <- max(0.001, leaf_lo_ratio)
leaf_hi_ratio <- min(0.200, leaf_hi_ratio)

PARAM$hiperparametertuning$hs_semillerio <- makeParamSet(
  makeIntegerParam("num_iterations",    lower = 15L,  upper = 60L),   # chico/rápido
  makeNumericParam("learning_rate",     lower = 0.10, upper = 0.35),
  makeNumericParam("feature_fraction",  lower = 0.20, upper = 0.60),
  makeNumericParam("coverage",          lower = 0.05, upper = 0.30),
  makeNumericParam("leaf_size",         lower = leaf_lo_ratio, upper = leaf_hi_ratio),
  makeNumericParam("bagging_fraction",  lower = 0.50, upper = 0.80),
  makeIntegerParam("bagging_freq",      lower = 1L, upper = 1L)
)

```

Función "señora caja negra"  que es llamada para verificar la realidad por la Bayesian Optimization

```r
# === Notificador (ntfy) queda igual si ya lo tenés ===
if (!exists("notificar")) {
  notificar <- function(msg) system(sprintf('curl -d "%s" https://ntfy.sh/juani_canal_cloud', msg))
}

`%||%` <- function(a,b) if (is.null(a)) b else a

mbo_progress_monitor <- function(path = "HT_RF.RDATA",
                                 maximize = TRUE,
                                 poll_sec = 30,
                                 eps = 1e-9) {
  best <- if (maximize) -Inf else Inf
  last_evals <- 0L
  last_improv_eval <- 0L

  repeat {
    Sys.sleep(poll_sec)
    if (!file.exists(path)) next

    env <- new.env(parent = emptyenv())
    ok <- tryCatch({ load(path, envir = env); TRUE }, error = function(e) FALSE)
    if (!ok || !exists("opt.state", envir = env)) next

    os <- get("opt.state", envir = env)
    df <- tryCatch(as.data.frame(os$opt.path), error = function(e) NULL)
    if (is.null(df) || nrow(df) == 0) next

    evals <- nrow(df)
    if (evals == last_evals) next
    last_evals <- evals

    # cuántos fueron parte del diseño inicial
    n_init <- os$control$init.design.points %||% 0L
    it_smart <- max(0L, evals - n_init)  # iteración "inteligente" efectiva
    it_total <- os$control$iters %||% NA_integer_

    # columna y/objetivo
    ycol <- if ("y" %in% names(df)) "y" else {
      yc <- grep("^y", names(df), ignore.case = TRUE, value = TRUE)
      if (length(yc)) yc[1] else next
    }
    yvec <- suppressWarnings(as.numeric(df[[ycol]]))
    if (!length(yvec)) next

    cur_best <- if (maximize) max(yvec, na.rm = TRUE) else min(yvec, na.rm = TRUE)
    improved <- if (maximize) (cur_best > best + eps) else (cur_best < best - eps)

    if (!is.finite(best) || improved) {
      best <- cur_best
      last_improv_eval <- evals
      # Notifico SIEMPRE la mejora, mostrando it inteligente (sin contar init)
      notificar(sprintf("⭐ Mejora: %.5f  | iter smart %d/%s (init=%d, eval=%d)",
                        cur_best, it_smart,
                        ifelse(is.na(it_total), "?", as.character(it_total)),
                        n_init, evals))
    } else {
      since <- evals - last_improv_eval
      if (since %% 20L == 0L && since > 0L) {
        notificar(sprintf("⏳ %d evals sin mejora | mejor=%.5f | smart=%d",
                          since, best, it_smart))
      }
    }
  }
}

```

```r
# En  x llegan los parmaetros de la bayesiana
#  devuelve la AUC en validate del modelo entrenado
#  en el parametro x llegan los hiperparámetros que se estan optimizando

EstimarGanancia_AUC_lightgbm <- function(x) {

  # x pisa (o agrega) a param_fijos
  param_completo <- modifyList(PARAM$lgbm$param_fijos, x)

  # ❌ Nunca cambiar seeds a nivel Dataset (rechazar si vinieron en x/fijos)
  param_completo$data_random_seed <- NULL
  param_completo$feature_fraction_seed <- NULL
  param_completo$bagging_seed <- NULL
    
  # ✅ Asegurar estocasticidad (si no lo seteó la OB)
  if (is.null(param_completo$bagging_freq))     param_completo$bagging_freq <- 1L
  if (is.null(param_completo$bagging_fraction)) param_completo$bagging_fraction <- 0.6
  if (is.null(param_completo$feature_fraction)) param_completo$feature_fraction <- 0.6
  param_completo$deterministic <- FALSE


  # hago la transformacion de leaf_size y  coverage
  if( "leaf_size"  %in% names(param_completo) &
      "coverage"  %in% names(param_completo)
  )
  {
    # primero defino el tamaño de las hojas
    param_completo$min_data_in_leaf <- pmax( 1,  round( nrow(dtrain) * param_completo$leaf_size )  )
    # luego la cantidad de hojas en funcion del valor anterior, el coverage, y la cantidad de registros
    param_completo$num_leaves <- pmin( 131072,
      pmax( 8,  round( ( param_completo$coverage * nrow( dtrain ) / param_completo$min_data_in_leaf ) ) ))
  }

  if( "leaf_size"  %in% names(param_completo) &
      !( "coverage"  %in% names(param_completo) )
  )
  {
    # primero defino el tamaño de las hojas
    param_completo$min_data_in_leaf <- pmax( 1,  round( nrow(dtrain) * param_completo$leaf_size )  )
  }


  # entreno LightGBM
  modelo_train <- lgb.train(
    data= dtrain,
    valids= list(valid = dvalidate),
    eval= "auc",
    param= param_completo,
    verbose= -100
  )

  # recupero la AUC en validation
  AUC <- modelo_train$record_evals$valid$auc$eval[[modelo_train$best_iter]]

  # esta es la forma de devolver un parametro extra
  attr(AUC, "extras") <- list("num_iterations"= modelo_train$best_iter)

  # hago espacio en la memoria
  rm(modelo_train)
  gc(full= TRUE, verbose= FALSE)

  message(format(Sys.time(), "%a %b %d %X %Y"), " AUC ", AUC)

  return(AUC)
}
```

```r
# === CHUNK 4: OB semillero-aware (promedio multisemilla + penalización por sd) ===
if (!exists("PARAM")) PARAM <- list()

PARAM$OB <- modifyList(
  list(
    seeds = c(100887, 200271, 333757, 567443, 875313),  # podés subir a 8-10 si querés
    alpha = 0.35,                             # objetivo = mean - alpha*sd
    usar_paramset_debil = TRUE                # TRUE: usa hs_semillerio, FALSE: hs_full
  ),
  if (!is.null(PARAM$OB)) PARAM$OB else list()
)

Estimar_AUC_seedavg_stable <- function(x) {
  aucs <- vapply(PARAM$OB$seeds, function(s) {
    set.seed(s, kind = "L'Ecuyer-CMRG")     # semilla global
    x2 <- modifyList(x, list(seed = s))     # sólo 'seed' al booster
    as.numeric(EstimarGanancia_AUC_lightgbm(x2))
  }, numeric(1))
  mean(aucs) - PARAM$OB$alpha * stats::sd(aucs)
}


```

```r
# === Normalizar opt.path y reconstruir pseudo-HP/HP de forma robusta ===
library(data.table)

# Convierte siempre a data.frame primero (evita rarezas de clases internas)
tb_bayesiana <- as.data.table(as.data.frame(bayesiana_salida$opt.path))

# 1) Arreglos de nombres/typos más comunes
if (!"num_iterations" %in% names(tb_bayesiana) && "num_interations" %in% names(tb_bayesiana)) {
  setnames(tb_bayesiana, "num_interations", "num_iterations")
}
if (!"y" %in% names(tb_bayesiana) && "perf" %in% names(tb_bayesiana)) {
  setnames(tb_bayesiana, "perf", "y")
}

# 2) Tamaño del train para mapear entre pseudo-HP y HP reales
nobs_train <- if (exists("dtrain")) nrow(dtrain) else if (exists("dataset")) nrow(dataset) else NA_integer_
if (is.na(nobs_train)) stop("No encuentro ni 'dtrain' ni 'dataset' para calcular nobs_train.")

# 3) Asegurá existencia de columnas destino (sin usarlas en 'i' si no existen)
for (nm in c("leaf_size","coverage","min_data_in_leaf","num_leaves")) {
  if (!nm %in% names(tb_bayesiana)) tb_bayesiana[, (nm) := NA_real_]
}

# 4A) Si EXISTEN min_data_in_leaf / num_leaves → completar leaf_size / coverage SOLO donde falten
if ("min_data_in_leaf" %in% names(tb_bayesiana)) {
  tb_bayesiana[is.na(leaf_size) & !is.na(min_data_in_leaf),
               leaf_size := pmax(1 / nobs_train, min_data_in_leaf / nobs_train)]
}
if (all(c("min_data_in_leaf","num_leaves") %in% names(tb_bayesiana))) {
  tb_bayesiana[is.na(coverage) & !is.na(min_data_in_leaf) & !is.na(num_leaves),
               coverage := (num_leaves * pmax(1, min_data_in_leaf)) / nobs_train]
}

# 4B) Si NO existen min_data_in_leaf / num_leaves pero SÍ leaf_size / coverage → reconstruir HP reales
tb_bayesiana[is.na(min_data_in_leaf) & !is.na(leaf_size),
             min_data_in_leaf := pmax(1, round(nobs_train * leaf_size))]

tb_bayesiana[is.na(num_leaves) & !is.na(coverage) & !is.na(min_data_in_leaf),
             num_leaves := pmin(131072L,
                                pmax(8L, round((coverage * nobs_train) / pmax(1, min_data_in_leaf))))]

# 5) Clamp a dominios conocidos (evita desbordes por redondeos)
tb_bayesiana[, leaf_size := pmin(pmax(leaf_size, 0.001), 0.200)]
tb_bayesiana[, coverage  := pmin(pmax(coverage , 0.050), 1.000)]

# 6) Redondeo / tipos prolijos
tb_bayesiana[, `:=`(
  leaf_size        = round(leaf_size, 6),
  coverage         = round(coverage , 6),
  min_data_in_leaf = as.integer(round(min_data_in_leaf)),
  num_leaves       = as.integer(round(num_leaves))
)]

# 7) Guardar log enriquecido
fwrite(tb_bayesiana, file = "BO_log_enriquecido.txt", sep = "\t")

# (Opcional) Mini resumen
cat("Evals:", nrow(tb_bayesiana), 
    " | Mejor y:", signif(if ("y" %in% names(tb_bayesiana)) max(tb_bayesiana$y, na.rm=TRUE) else NA_real_, 6),
    " | Ult. mod:", as.character(file.info("BO_log_enriquecido.txt")$mtime), "\n")
```

```r
# almaceno los resultados de la Bayesian Optimization
# y capturo los mejores hiperparametros encontrados
# almaceno los resultados...
# === RESULTADOS OB: normalización + log + mejor HP (pseudo-HP) ===
library(data.table)

# Helper: estandariza el opt.path y crea coverage/leaf_size si faltan
normalizar_opt_path <- function(opt.path, nobs_train) {
  tb <- as.data.table(opt.path)

  # 1) Fix posibles typos / columnas
  if (!"num_iterations" %in% names(tb) && "num_interations" %in% names(tb)) {
    tb[, num_iterations := num_interations]
  }
  if (!"y" %in% names(tb) && "perf" %in% names(tb)) setnames(tb, "perf", "y")

  # 2) Asegurar que existan columnas pseudo-HP
  if (!"leaf_size" %in% names(tb)) tb[, leaf_size := NA_real_]
  if (!"coverage"  %in% names(tb)) tb[, coverage  := NA_real_]

  # 3) Reconstruir pseudo-HP desde HP directos (si existen)
  #    min_data_in_leaf = nobs * leaf_size
  #    coverage = num_leaves * min_data_in_leaf / nobs
  if ("min_data_in_leaf" %in% names(tb) && is.finite(nobs_train) && nobs_train > 0) {
    tb[is.na(leaf_size) & !is.na(min_data_in_leaf),
       leaf_size := pmax(0.001, pmin(0.200, min_data_in_leaf / nobs_train))]
  }
  if (all(c("num_leaves","min_data_in_leaf") %in% names(tb)) && is.finite(nobs_train) && nobs_train > 0) {
    tb[is.na(coverage) & !is.na(num_leaves) & !is.na(min_data_in_leaf),
       coverage := pmax(0.050, pmin(1.000, (num_leaves * min_data_in_leaf) / nobs_train))]
  }

  # 4) Prolijidad
  tb[, `:=`(leaf_size = round(leaf_size, 6),
            coverage  = round(coverage , 6))]

  tb
}

# --- Construir tabla estandarizada desde el opt.path ---
nobs_train <- if (exists("dtrain")) nrow(dtrain) else if (exists("dataset")) nrow(dataset) else NA_integer_
if (!is.finite(nobs_train) || nobs_train <= 0) stop("No pude determinar nobs_train desde dtrain/dataset.")

tb_bayesiana <- normalizar_opt_path(bayesiana_salida$opt.path, nobs_train)

# Orden y guardado del log (incluye coverage/leaf_size)
setorder(tb_bayesiana, -y, -num_iterations)
fwrite(tb_bayesiana, file = "BO_log.txt", sep = "\t")

# --- Chequeo de puntos únicos (para entender los warnings del surrogate) ---
cols_hp_direct <- intersect(c("num_iterations","learning_rate","feature_fraction",
                               "num_leaves","min_data_in_leaf","bagging_fraction","bagging_freq"),
                             names(tb_bayesiana))
puntos_unicos <- uniqueN(tb_bayesiana[, ..cols_hp_direct])
cat(sprintf("📊 Puntos únicos en el surrogate: %d\n", puntos_unicos))
print(tail(unique(tb_bayesiana[, ..cols_hp_direct], fromLast = TRUE), 10))

# --- Mejor set de HP en formato pseudo-HP (lo que espera tu pipeline) ---
PARAM$out$lgbm$mejores_hiperparametros <- tb_bayesiana[
  1L,
  .(learning_rate, feature_fraction, coverage, leaf_size)
]
print(PARAM$out$lgbm$mejores_hiperparametros)
```
